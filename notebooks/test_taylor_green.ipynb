# Test: Taylor–Green vortex post-processing

Test notebook for the Post-processing section. It uses the analytic 2D Taylor–Green solution

$$u = \sin x \cos y \, e^{-2\nu t}, \qquad v = -\cos x \sin y \, e^{-2\nu t}$$

in place of solver output.

```
import numpy as np
import matplotlib.pyplot as plt

nu = 0.01
n = 64
x = np.linspace(0, 2*np.pi, n)
X, Y = np.meshgrid(x, x)
```

## Velocity field at t = 0

```
u = np.sin(X) * np.cos(Y)
v = -np.cos(X) * np.sin(Y)

fig, ax = plt.subplots(figsize=(5, 5))
c = ax.contourf(X, Y, np.hypot(u, v), levels=20, cmap='viridis')
ax.streamplot(X, Y, u, v, color='white', density=1.0, linewidth=0.6)
fig.colorbar(c, ax=ax, label='|u|')
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_aspect('equal')
plt.show()
```

## Kinetic energy decay

The domain-averaged kinetic energy should decay as $E(t) = E_0 e^{-4\nu t}$.

```
t = np.linspace(0, 50, 101)
E = np.array([0.5*np.mean((u**2 + v**2) * np.exp(-4*nu*ti)) for ti in t])

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(t, E/E[0], label='computed')
ax.plot(t[::10], np.exp(-4*nu*t[::10]), 'o', label='analytic')
ax.set_xlabel('t'); ax.set_ylabel('E / E0'); ax.legend(); ax.grid(alpha=0.3)
plt.show()

print(f'Max relative error: {np.max(np.abs(E/E[0] - np.exp(-4*nu*t))):.2e}')
```
